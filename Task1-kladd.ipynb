```python
def find_encoding(file):
    try:
        with open(file, encoding= 'utf-8') as f:
            f.read()
        Etype = 'utf-8'
    
    except UnicodeDecodeError:
        Etype = 'ISO-8859-1/ Latin 1'

    for file in Path('2026').rglob('*.log'):
        return (file, Etype)

        


find_encoding('C:/Users/danie/Documents/NMBU 4.år/Inf 201/Deliverables/W43_44/INF201_Deliverable_2/2026/library_cafe/POS_01/register_sales.log')

```
